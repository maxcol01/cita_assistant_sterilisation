```python
import pandas as pd
from pathlib import Path
path_to_doc = Path(".") / "documents"
```

```python
row = {}
documents = []
for item in path_to_doc.iterdir():
    row["name"] = item.name
    row["location"] = item.as_posix()
    documents.append(row.copy())
    db = pd.DataFrame(documents)
```
