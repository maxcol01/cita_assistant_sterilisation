```python
import pandas as pd
from pathlib import Path
path_to_doc = Path(".") / "documents"
```

```python
from datetime import datetime
```

```python

documents = []
for item in path_to_doc.iterdir():
    row = {}
    if item.suffix == ".pdf":
        print(item.as_posix())
        row["name"] = item.name
        row["location"] = item.as_posix()
        row["date"] = datetime.today()
        documents.append(row)
db = pd.DataFrame(documents)
```
